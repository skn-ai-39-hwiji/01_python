# Type Hint
- python은 동적 타이핑 언어
    -> 변수에 저장되는 값에 따라서 자료형이 결정된다

- type hint는 코드 작성시 IDE(통합 개발 환경)과 정적 분석 도구가 데이터 타입을 추론하고 경고할 수 있도록 돕는 표시

- 중요한점은 type hint 가 python 런타임에서 타입이 틀렸다고 에러를 발생시키지는 않는다.(경고만 한다!)



## 1. 동적 타이핑
- python에서는 같은 변수에 여러가지 서로 다른타입의 변수를 대입할 수 있다.
- 대입되는 값에 따라서 자료형이 결정된다.

````python
x = '안녕'
print(x, type(x))

x = 100
print(x, type(x))
````

## 2. 변수의 타입 힌트

```python
변수명: 타입 = 값
```
- `: 타입 ` == 변수에 대입할 값의 타입 힌트

````python
greeting: str = 'hello'
print(greeting, type(greeting))

greeting = 12345 # 경고발생 ,하지만 장상실행가능
print(greeting , type(greeting))

num: int = 100
pi: float = 3.141592
is_true: bool = True
lst: list = []

````

## 3. 컬렉션(자료구조 모음) 타입 힌트
- list, tuple, dict, set 같은 컬렉션에도 힌트 적용 가능
```python
변수명: 컬렉션[타입] = 값
```
- list[int]: 정수를 담는 리스트
- tuple[int, str, str]: 첫 번째는 정수, 두 번째와 세 번째는 문자열인 튜플
- dict[str, str | int]: key는 문자열, value는 문자열 또는 정수인 딕셔너리
- set[str]: 문자열을 담는 집합

````python
names: list[str] = ['홍길동','고길동']
user: tuple[int, str, list[str]] = ( 10 , "user1" , [1,2,3])

#key는 str
#value는 str 또는 int 또는 float
info: dict[str, str | int | float] = {
    "a": "abc",
    "b": 10,
    "c": 3.14
}

chars: set[str] = {'a','b','c'}
````
